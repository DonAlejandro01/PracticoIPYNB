```python
import pandas as pd
Datos = pd.read_csv('Horario - Horario.csv') # Carga de archivo CSV
print(Datos)
```

```python
Datos = Datos.fillna("Libre")
Datos.head()
```

```python
# Grafico de baras por cantidad de funcionarios operativos por dias
import matplotlib.pyplot as plt
SoloOperativos = Datos[Datos['Dia_Libre']== 'NO']
SoloOperativos.groupby('Fecha').size().plot(kind='bar')
ax = SoloOperativos.groupby("Fecha")["Empleado"].count().plot(kind="bar")

# Agregar números encima de cada barra
for i, v in enumerate(SoloOperativos.groupby("Fecha")["Empleado"].count()):
    ax.text(i, v + 0.1, str(v), ha='center')

plt.xlabel('Fecha')
plt.ylabel('Operativos por dia')
plt.title('Cantidad de funcionarios operativos por dia')
plt.show()  
print(Datos)
```
