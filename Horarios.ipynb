# Practico 1
El enfoque de este proyecto sera que a posterior se pueda generar o proyectar horarios para operarios del mundo de carga, el cual tienen que cumplir restricciones como horarios de trabajo, descansos, turnos, etc. Para esto se va a generar un dataset con datos ficticios de operarios, horarios, turnos y restricciones. Luego se aplicaran algoritmos de optimización para generar horarios que cumplan con las restricciones establecidas.

## Criterio de exito
- Cargar archivo CSV ✖️
- Lectura de datos ✖️
- Mostrar tabla de datos y grafico de barra por cantidad de funcionarios operativos por dia ✖️
